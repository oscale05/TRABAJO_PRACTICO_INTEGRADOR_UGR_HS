````python
#@title Ejercicio 02.2: Separar imágenes en plates y completes
import os
from pathlib import Path

imgs_dir = Path("port_log/data/raw/imgs/port_log_images")
if not imgs_dir.exists():
    imgs_dir = Path("port_log/data/raw/imgs")

image_files = list(imgs_dir.rglob("*.jpg")) + list(imgs_dir.rglob("*.png")) + list(imgs_dir.rglob("*.jpeg"))

plates = []
completes = []

for img_path in sorted(image_files):
    filename = img_path.name.lower()
    if "plate" in filename:
        plates.append(img_path)
    else:
        completes.append(img_path)

print(f"El grupo 'plates' contiene {len(plates)} imágenes.")
print(f"El grupo 'completes' contiene {len(completes)} imágenes.")
print(f"Criterio de separación: nombre de archivo contiene 'plate' -> plates, resto -> completes")

# Guardar listas para siguiente celda
import json
with open('plates_list.json', 'w') as f:
    json.dump([str(p) for p in plates], f)
with open('completes_list.json', 'w') as f:
    json.dump([str(p) for p in completes], f)
````

````python
#@title Ejercicio 02.3: Construir group_images y guardar JSON
import json
from pathlib import Path
from PIL import Image

with open('plates_list.json', 'r') as f:
    plates = [Path(p) for p in json.load(f)]
with open('completes_list.json', 'r') as f:
    completes = [Path(p) for p in json.load(f)]

def get_image_info(img_path):
    try:
        with Image.open(img_path) as img:
            width, height = img.size
    except:
        width, height = 0, 0
    
    return {
        "filename": img_path.name,
        "width": width,
        "height": height,
        "area": width * height,
        "path": str(img_path),
        "matricula_imagen": ""
    }

group_images = {
    "plates": [get_image_info(p) for p in plates],
    "completes": [get_image_info(c) for c in completes]
}

output_path = Path("port_log/data/interim/group_images.json")
output_path.parent.mkdir(parents=True, exist_ok=True)
with open(output_path, 'w') as f:
    json.dump(group_images, f, indent=2)

print(f"group_images guardado en: {output_path}")
print(f"Plates: {len(group_images['plates'])}, Completes: {len(group_images['completes'])}")
````

````python
#@title Ejercicio 02.4: Stats por grupo (resolución, área, tamaño promedio)
import json
from pathlib import Path

with open('port_log/data/interim/group_images.json', 'r') as f:
    group_images = json.load(f)

for grupo in ['plates', 'completes']:
    imgs = group_images[grupo]
    if not imgs:
        continue
    
    total_width = sum(img['width'] for img in imgs)
    total_height = sum(img['height'] for img in imgs)
    total_area = sum(img['area'] for img in imgs)
    total_size_kb = sum(Path(img['path']).stat().st_size / 1024 for img in imgs)
    n = len(imgs)
    
    avg_res = f"{total_height // n} x {total_width // n}"
    avg_area = total_area // n
    avg_size_kb = total_size_kb / n
    
    print(f"El grupo '{grupo}' posee:")
    print(f"- resolución promedio: {avg_res}")
    print(f"- área promedio {avg_area} en pixeles")
    print(f"- tamaño promedio {avg_size_kb:.2f} kB")
    print()
````

- Definir una función que muestra imágenes aleatorias de cada grupo en grilla de 2 columnas. Cada celda debe mostrar el nombre del archivo y las dimensiones. Llamar a la función mostrando como máximo 4 imágenes de cada grupo.

```
def mostrar_muestra(gropo_imagenes: Any, cantidad_mostrar: int = 4):
  """ Muestra una cantidad de imágenes por pantalla/notebook.

  Args:
    grupo_imagenes (Any): Conjunto de imágenes a mostrar.
    cantidad_mostrar (int): Cantidad de imágenes a mostrar del grupo que se pasa como parámetro. Por defecto 4.
  """
```

````python
#@title Ejercicio 02.5: Función mostrar_muestra (grilla 2 columnas)
import json
import random
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

with open('port_log/data/interim/group_images.json', 'r') as f:
    group_images = json.load(f)

def mostrar_muestra(grupo_imagenes, cantidad_mostrar=4):
    """ Muestra una cantidad de imágenes por pantalla/notebook.

    Args:
        grupo_imagenes (list): Lista de diccionarios con info de imágenes.
        cantidad_mostrar (int): Cantidad de imágenes a mostrar. Por defecto 4.
    """
    if not grupo_imagenes:
        print("No hay imágenes en este grupo")
        return
    
    muestra = random.sample(grupo_imagenes, min(cantidad_mostrar, len(grupo_imagenes)))
    
    cols = 2
    rows = (len(muestra) + cols - 1) // cols
    
    fig, axes = plt.subplots(rows, cols, figsize=(10, 5 * rows))
    if rows == 1:
        axes = axes.reshape(1, -1)
    axes = axes.flatten()
    
    for idx, img_info in enumerate(muestra):
        try:
            with Image.open(img_info['path']) as img:
                axes[idx].imshow(img)
                axes[idx].set_title(f"{img_info['filename']}\n{img_info['width']}x{img_info['height']}")
                axes[idx].axis('off')
        except Exception as e:
            axes[idx].text(0.5, 0.5, f"Error cargando:\n{e}", ha='center', va='center')
            axes[idx].axis('off')
    
    # Hide unused subplots
    for idx in range(len(muestra), len(axes)):
        axes[idx].axis('off')
    
    plt.tight_layout()
    plt.show()

# Llamar a la función para cada grupo
print("=== MUESTRA PLATES ===")
mostrar_muestra(group_images['plates'], 4)

print("=== MUESTRA COMPLETES ===")
mostrar_muestra(group_images['completes'], 4)
````

- **Escala de grises**: convertir todas las imágenes originales y guardar en:
  - `port_log/data/interim/imgs/03_01_gray/plates`
  - `port_log/data/interim/imgs/03_01_gray/completes`



````python
#@title Ejercicio 03.1: Escala de grises
import json
from pathlib import Path
import cv2

with open('port_log/data/interim/group_images.json', 'r') as f:
    group_images = json.load(f)

output_base = Path('port_log/data/interim/imgs/03_01_gray')
for grupo in ['plates', 'completes']:
    out_dir = output_base / grupo
    out_dir.mkdir(parents=True, exist_ok=True)
    
    for img_info in group_images[grupo]:
        img = cv2.imread(img_info['path'])
        if img is None:
            continue
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        out_path = out_dir / img_info['filename']
        cv2.imwrite(str(out_path), gray)
        img_info['path_gray'] = str(out_path)

# Actualizar group_images con paths de grises
with open('port_log/data/interim/group_images.json', 'w') as f:
    json.dump(group_images, f, indent=2)

print('Escala de grises completada. Imagenes guardadas en 03_01_gray/{plates,completes}/')

# Visualizar muestra
mostrar_muestra(group_images['plates'][:4], 4)
mostrar_muestra(group_images['completes'][:4], 4)
````

- **Ecualización de histograma**: aplicar sobre las imágenes en escala de grises para mejorar el contraste (útil para imágenes nocturnas o subexpuestas). Guardar en:
  - `port_log/data/interim/imgs/03_02_equalized/plates`
  - `port_log/data/interim/imgs/03_02_equalized/completes`



````python
#@title Ejercicio 03.2: Ecualizacion de histograma
import json
from pathlib import Path
import cv2

with open('port_log/data/interim/group_images.json', 'r') as f:
    group_images = json.load(f)

output_base = Path('port_log/data/interim/imgs/03_02_equalized')
for grupo in ['plates', 'completes']:
    out_dir = output_base / grupo
    out_dir.mkdir(parents=True, exist_ok=True)
    
    for img_info in group_images[grupo]:
        if 'path_gray' not in img_info:
            continue
        gray = cv2.imread(img_info['path_gray'], cv2.IMREAD_GRAYSCALE)
        if gray is None:
            continue
        equalized = cv2.equalizeHist(gray)
        out_path = out_dir / img_info['filename']
        cv2.imwrite(str(out_path), equalized)
        img_info['path_equalized'] = str(out_path)

# Actualizar group_images
with open('port_log/data/interim/group_images.json', 'w') as f:
    json.dump(group_images, f, indent=2)

print('Ecualizacion completada. Imagenes guardadas en 03_02_equalized/{plates,completes}/')

# Visualizar muestra
mostrar_muestra(group_images['plates'][:4], 4)
mostrar_muestra(group_images['completes'][:4], 4)
````


- **Suavizado**: aplicar blur gaussiano sobre las imágenes ecualizadas. Guardar en:
  - `port_log/data/interim/imgs/03_03_blur/plates`
  - `port_log/data/interim/imgs/03_03_blur/completes`


````python
#@title Ejercicio 03.3: Suavizado (Blur Gaussiano)
import json
from pathlib import Path
import cv2

with open('port_log/data/interim/group_images.json', 'r') as f:
    group_images = json.load(f)

output_base = Path('port_log/data/interim/imgs/03_03_blur')
for grupo in ['plates', 'completes']:
    out_dir = output_base / grupo
    out_dir.mkdir(parents=True, exist_ok=True)
    
    for img_info in group_images[grupo]:
        if 'path_equalized' not in img_info:
            continue
        equalized = cv2.imread(img_info['path_equalized'], cv2.IMREAD_GRAYSCALE)
        if equalized is None:
            continue
        # Kernel 5x5, sigma=1.0
        blurred = cv2.GaussianBlur(equalized, (5, 5), 1.0)
        out_path = out_dir / img_info['filename']
        cv2.imwrite(str(out_path), blurred)
        img_info['path_blurred'] = str(out_path)

# Actualizar group_images
with open('port_log/data/interim/group_images.json', 'w') as f:
    json.dump(group_images, f, indent=2)

print('Blur gaussiano completado. Imagenes guardadas en 03_03_blur/{plates,completes}/')

# Visualizar muestra
mostrar_muestra(group_images['plates'][:4], 4)
mostrar_muestra(group_images['completes'][:4], 4)
````

````python
#@title Función extraer_matricula

import easyocr

def extraer_matricula(image_path: str) -> str:
  """Extrae la matrícula de un buque desde una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Texto detectado o cadena vacía si no hay detección.
  """
  reader = easyocr.Reader(['en'])
  detection: list = reader.readtext(image_path)
  if not detection:
    return ''
  return ''.join(
    d[1] for d in detection
    if isinstance(d, tuple) and len(d) == 3
    and isinstance(d[1], str)
  )
````

````python
#@title Ejercicio 04: Extraccion de matriculas (OCR) y matching con dataset
import json
import pandas as pd
from pathlib import Path

# Cargar group_images
with open('port_log/data/interim/group_images.json', 'r') as f:
    group_images = json.load(f)

# Cargar dataset de movimientos
df_mov = pd.read_csv('port_log/data/interim/port_movements.csv')

# Preparar matriculas del dataset para matching (solo alfanumericos)
def normalizar_matricula(mat):
    if pd.isna(mat):
        return ''
    return ''.join(c for c in str(mat).upper() if c.isalnum())

df_mov['matricula_norm'] = df_mov['matricula'].apply(normalizar_matricula)
matriculas_dataset = df_mov['matricula_norm'].tolist()

# Inicializar EasyOCR reader una sola vez
reader = easyocr.Reader(['en'])

def extraer_matricula_optimizada(image_path: str) -> str:
    """Extrae la matricula usando EasyOCR ya inicializado."""
    try:
        detection = reader.readtext(image_path)
        if not detection:
            return ''
        textos = [d[1] for d in detection if isinstance(d, tuple) and len(d) == 3 and isinstance(d[1], str)]
        return ''.join(textos).upper()
    except Exception as e:
        print(f"Error OCR en {image_path}: {e}")
        return ''

# Funcion de matching: comparar alfanumericos ignorando guiones/espacios
def calcular_ratio(match_original, match_detectada):
    """Calcula porcentaje de coincidencia posicion por posicion."""
    if not match_original or not match_detectada:
        return 0.0
    min_len = min(len(match_original), len(match_detectada))
    if min_len == 0:
        return 0.0
    coincidencias = sum(1 for i in range(min_len) if match_original[i] == match_detectada[i])
    return (coincidencias / len(match_original)) * 100

# Extraer matriculas de todas las imagenes
print("Extrayendo matriculas con OCR...")
for grupo in ['plates', 'completes']:
    for img_info in group_images[grupo]:
        if 'path_canny' in img_info:
            matricula = extraer_matricula_optimizada(img_info['path_canny'])
        else:
            matricula = extraer_matricula_optimizada(img_info['path'])
        matricula_clean = ''.join(c for c in matricula if c.isalnum())
        img_info['matricula_imagen'] = matricula_clean
        print(f"  {img_info['filename']}: '{matricula_clean}'")

# Actualizar group_images con matriculas extraidas
with open('port_log/data/interim/group_images.json', 'w') as f:
    json.dump(group_images, f, indent=2)

print("\nMatriculas extraidas y guardadas en group_images.json")

# Matching con dataset
print("\nRealizando matching con dataset...")
matches = []

for grupo in ['plates', 'completes']:
    for img_info in group_images[grupo]:
        matricula_ocr = img_info.get('matricula_imagen', '')
        if not matricula_ocr:
            continue
        
        best_match = None
        best_ratio = 0.0
        best_idx = -1
        
        for idx, mat_dataset in enumerate(matriculas_dataset):
            if not mat_dataset:
                continue
            ratio = calcular_ratio(mat_dataset, matricula_ocr)
            if ratio > best_ratio:
                best_ratio = ratio
                best_match = mat_dataset
                best_idx = idx
        
        if best_ratio >= 75.0 and best_idx >= 0:
            row = df_mov.iloc[best_idx]
            matches.append({
                'movimiento_id': row['movimiento_id'],
                'imagen': img_info['path'],
                'matricula_imagen': matricula_ocr,
                'ratio': round(best_ratio, 2),
                'grupo_imagen': grupo,
                'matricula_original': row['matricula']
            })
            print(f"  MATCH: {row['movimiento_id']} | OCR: {matricula_ocr} | Dataset: {row['matricula']} | Ratio: {best_ratio:.1f}% | Grupo: {grupo}")
        else:
            print(f"  NO MATCH: {img_info['filename']} | OCR: {matricula_ocr} | Mejor: {best_match} ({best_ratio:.1f}%)")

# Crear DataFrame resultado
df_result = pd.DataFrame(matches)

if len(df_result) > 0:
    cols_extra = ['buque_id', 'fecha_ingreso', 'hora_ingreso', 'muelle', 'tipo_carga', 'origen', 'velocidad_ingreso', 'velocidad_maxima_muelle', 'exceso_velocidad', 'exceso_velocidad_real', 'duracion_horas', 'estado_despacho']
    for col in cols_extra:
        if col in df_mov.columns:
            mapa = df_mov.set_index('movimiento_id')[col].to_dict()
            df_result[col] = df_result['movimiento_id'].map(mapa)

output_path = Path('port_log/data/processed/port_movements_image.csv')
output_path.parent.mkdir(parents=True, exist_ok=True)
df_result.to_csv(output_path, index=False)

print(f"\n=== RESULTADO ===")
print(f"Total matches encontrados: {len(df_result)}")
print(f"Guardado en: {output_path}")
if len(df_result) > 0:
    print(df_result[['movimiento_id', 'matricula_original', 'matricula_imagen', 'ratio', 'grupo_imagen']].to_string(index=False))
else:
    print("No se encontraron matches con ratio >= 75%")
````
